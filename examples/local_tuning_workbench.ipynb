# 学習データ内で Ridge の条件を選ぶ

保存した Iceberg 学習データ48行だけで、alpha 0.1・1・10 を固定3分割で比較します。平均 validation MSE で選び、同点なら小さい alpha を採用し、48行すべてで再学習します。9回の検証と1回の再学習を行います。

既視の holdout 16行は選択後の参考評価だけに使用します。独立した最終テストではありません。MLflow登録・モデル昇格は行いません。初回Run Allは学習を実行しません。

```python
from IPython.display import HTML, display
from tools.local_experiment.notebook_lab.tune import tune, recover_tuning, render_tuning
from tools.local_experiment.notebook.client import NotebookError
```

## 保存元を指定して実行する

保存済み Iceberg 学習の `training_token` を指定し、`run_tuning = True` に変えて実行セルを動かします。alpha・分割数・seed は固定です。実行セルをもう一度動かすと新しい10回の学習が始まるため、完了後の再表示には下の保存実行トークンを使ってください。

```python
training_token = ""
run_tuning = False
tuning_result = None
tuning_html = None
```

```python
tuning_result = None
tuning_html = None
candidate = None
candidate_html = None
if run_tuning:
    try:
        candidate = tune(training_token)
        candidate_html = render_tuning(candidate)
    except NotebookError as error:
        candidate = None
        candidate_html = None
        print("CVを完了できませんでした。保存証跡と未完了の実行を確認してください:", str(error))
    except BaseException:
        candidate = None
        candidate_html = None
        raise
    else:
        tuning_result = candidate
        tuning_html = candidate_html
        display(HTML(tuning_html))
```

## 保存結果を再表示する

成功結果の `attempt_token` を `tuning_attempt` に指定します。当時の入力・結果・終了証跡と保存元モデルを照合します。追加の学習やモデル登録は行いません。

```python
tuning_attempt = ""
saved_tuning = None
saved_tuning_html = None
saved_candidate = None
saved_candidate_html = None
if tuning_attempt:
    try:
        saved_candidate = recover_tuning(tuning_attempt)
        saved_candidate_html = render_tuning(saved_candidate)
    except NotebookError as error:
        saved_candidate = None
        saved_candidate_html = None
        print("保存されたCVは確認できませんでした:", str(error))
    except BaseException:
        saved_candidate = None
        saved_candidate_html = None
        raise
    else:
        saved_tuning = saved_candidate
        saved_tuning_html = saved_candidate_html
        display(HTML(saved_tuning_html))
```
