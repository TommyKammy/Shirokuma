# 保存データで Ridge の条件を比較する

保存済みの Iceberg 学習を選び、alpha を変更して1回学習します。保存モデルは alpha=1 です。
同じ48行を学習に使い、同じ16行の holdout を再利用する探索的比較です。独立した最終評価ではありません。新しい Reader、MLflow run、LLM 呼出しはありません。

```python
from IPython.display import HTML, JSON, display
from tools.local_experiment.mlflow.training_runs import collect
from tools.local_experiment.notebook_lab.compare import compare, render_comparison
inventory = collect()
iceberg_runs = [run for run in inventory['runs'] if run['kind'] == 'iceberg']
display(JSON(inventory))
```

下の token を一覧に表示された `training_token` に置き換えてください。alpha は 0.001〜1000 の有限数値です。学習は専用の既存 ML 環境で実行します。

```python
training_token = ""
alpha = 10.0
```

```python
if inventory['result'] != 'complete':
    print("学習の一覧を確認できませんでした。最初のセルの結果を確認してください。")
elif training_token not in {run['training_token'] for run in iceberg_runs}:
    print("一覧の kind が iceberg の training_token を入力してください。")
else:
    comparison = compare(training_token, alpha)
    display(HTML(render_comparison(comparison)))
```

比較表の予測差は「比較モデル − 保存モデル」です。小さい RMSE・MAE、大きい R² がこの保存データ上での良い値です。自動でモデルを選びません。
