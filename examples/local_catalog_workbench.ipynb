# CatalogとLineageを読む

固定fixture `smoke.fixture_v1` の所有者・説明・列・記録された1 hopの関連を確認します。Catalogには過去の関連が残る場合があります。現在のSQLテーブルの実在確認やSQL実行は行いません。

初回Run Allは未選択のため、Catalog・モデルを呼びません。

```python
from IPython.display import HTML, display
from tools.local_experiment.notebook_lab.query import catalog, recover_catalog, render_catalog
from tools.local_experiment.notebook.client import NotebookError
```

## 対象を指定して読む

`table = "smoke.fixture_v1"` と `run_catalog = True` を設定し、次の実行セルを動かします。この固定fixtureのみを受け付けます。1回の認証と最大12 GETで保存情報を確認し、結果を表示します。

```python
table = ""
run_catalog = False
catalog_result = None
catalog_html = None
```

```python
catalog_result = None
catalog_html = None
candidate = None
candidate_html = None
if run_catalog:
    try:
        candidate = catalog(table)
        candidate_html = render_catalog(candidate)
    except NotebookError as error:
        candidate = None
        candidate_html = None
        print("Catalogは確認できませんでした。対象と保存証跡を確認してください:", str(error))
    except BaseException:
        candidate = None
        candidate_html = None
        raise
    else:
        catalog_result = candidate
        catalog_html = candidate_html
        display(HTML(catalog_html))
```

## 保存結果を再表示する

成功したCatalog実行の `attempt_token` を指定します。保存された当時の結果と証跡を照合し、追加GETなしで表示します。現在の状態を読み直す操作ではありません。

```python
catalog_attempt = ""
saved_catalog = None
saved_catalog_html = None
saved_candidate = None
saved_candidate_html = None
if catalog_attempt:
    try:
        saved_candidate = recover_catalog(catalog_attempt)
        saved_candidate_html = render_catalog(saved_candidate)
    except NotebookError as error:
        saved_candidate = None
        saved_candidate_html = None
        print("保存されたCatalogは確認できませんでした:", str(error))
    except BaseException:
        saved_candidate = None
        saved_candidate_html = None
        raise
    else:
        saved_catalog = saved_candidate
        saved_catalog_html = saved_candidate_html
        display(HTML(saved_catalog_html))
```
