# ローカル Qwen でテーブルをプレビュー
明示した iceberg テーブルの列・件数を自然文で指定します。`filtered=True` は型付き単一等値条件です。Qwen は schema と依頼だけを読み、行・SQL・Notebook コードは生成しません。
各実行は新しい依頼です。前の実行が終わってから次を実行してください。取消は Interrupt を1回使い、所有 worker の後片付けを待ってください。Notebook の入力と表示行は専用 workspace に保存されます。

```python
from IPython.display import HTML, display
from tools.local_experiment.notebook_lab.query import preview, render_preview
from tools.local_experiment.notebook.client import NotebookError
```

```python
table = "smoke.fixture_v1"  # catalog は iceberg 固定
request = "nameとidを1件表示"
filtered = False  # 等値条件の例: True / idが1の行からnameとidを1件表示
```

```python
result = None
try:
    result = preview(table, request, filtered=filtered)
except NotebookError as error:
    print("実行を完了できませんでした。固定エラーコード:", str(error))
    print("未確認の実行は自動再試行されません。詳細は専用 queries ディレクトリを確認してください。")
else:
    display(HTML(render_preview(result)))
```
