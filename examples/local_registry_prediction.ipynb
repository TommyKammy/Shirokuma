# 登録したモデルで予測する
明示したname/versionを読取専用で照合し、保存済みRidgeモデルで入力を予測します。新しい学習・登録・Servingは行いません。初期状態のRun Allはversion未指定のためDB読取も予測も行いません。


```python
prediction = None
prediction_html = None
from IPython.display import HTML, display
from tools.local_experiment.notebook.client import NotebookError
from tools.local_experiment.notebook.registry import predict_registered, render_prediction
```

## モデルを選ぶ
登録済みversionの正確な番号を指定してください。aliasやstageは使用しません。


```python
model_name = "shirokuma-local-ridge"
model_version = ""  # 登録済みの version 番号を明示してください
prediction = None
prediction_html = None
```

## 入力を予測する
`x1`・`x2`・`x3`の有限数値を1〜64行指定します。セルの再実行ごとにversionと保存モデルの出典を照合し、入力順に予測します。新入力の正解精度を保証するものではありません。


```python
prediction_rows = [
    {"x1": 0.0, "x2": 0.0, "x3": 0.0},
    {"x1": 0.25, "x2": -0.75, "x3": 1.5},
]
prediction = None
prediction_html = None
candidate = None
candidate_html = None
if not model_version:
    print("登録済みモデルの version 番号を上のセルで指定してください。")
else:
    try:
        candidate = predict_registered(model_name, model_version, {"schema_version": 1, "rows": prediction_rows})
        candidate_html = render_prediction(candidate)
    except NotebookError as error:
        candidate = None
        candidate_html = None
        print("予測を確認できませんでした。version と入力を確認してください。固定エラーコード:", str(error))
    except BaseException:
        candidate = None
        candidate_html = None
        raise
    else:
        prediction = candidate
        prediction_html = candidate_html
        display(HTML(prediction_html))
```
