# 保存した比較を MLflow に記録する

一覧から選んだ候補の alpha・評価値・元の学習との関係を記録します。比較モデルをもう一度学習する必要はありません。


```python
from IPython.display import JSON, display
from tools.local_experiment.mlflow.comparison_runs import collect
from tools.local_experiment.notebook_lab.record_comparison import record_comparison
inventory = None
recording = None
try:
    inventory = collect()
except Exception:
    print("比較一覧を確認できませんでした。保存された比較の状態を確認してください。")
else:
    display(JSON(inventory))
```

一覧の runs にある comparison_token を下のセルへコピーしてください。初回の記録中は MLflow 画面が一時的に使えなくなり、終了後に戻ります。


```python
comparison_token = ""
```

```python
recording = None
if inventory is None or inventory["result"] != "complete":
    print("比較一覧を確認できませんでした。最初のセルを確認してください。")
elif comparison_token not in {run["comparison_token"] for run in inventory["runs"]}:
    print("一覧の runs に表示された comparison_token を入力してください。")
else:
    try:
        recording = record_comparison(comparison_token)
    except Exception:
        print("記録を確認できませんでした。保存された実行状態を確認してください。")
    else:
        print("保存された登録結果を表示します。" if recording["reused"] else "MLflow への記録を完了しました。")
        print("候補の alpha:", recording["comparison"]["candidate"]["alpha"])
        display(JSON(recording["import_receipt"]))
```

[MLflow の実験一覧を開く](http://127.0.0.1:18550/) → `shirokuma-local-ridge-comparisons-v1` を選ぶと、候補・元モデル・差分の評価値を確認できます。

同じ比較の再実行では、保存された登録結果を表示します。新しい run は追加しません。保存結果の再表示は、MLflow データベースの最新状態の再検証ではありません。

同じ holdout を使った探索的比較です。モデルを自動で採用・公開する操作はありません。Notebook を保存すると入力と表示結果を残せます。
