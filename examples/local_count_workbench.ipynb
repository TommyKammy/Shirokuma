# ローカル Qwen でテーブルの件数を数える
指定したテーブルの現在の snapshot に対し、全行または単一の型付き等値条件に一致する件数を取得します。Qwen が読むのは schema と依頼だけです。SQL・行・Notebook コードは生成しません。
集計は全 snapshot を読む場合があります。各実行は新しい依頼です。前の実行が終わってから次を実行し、取消は Interrupt を1回使って後片付けを待ってください。入力と集計結果は専用 workspace に保存されます。

```python
from IPython.display import HTML, display
from tools.local_experiment.notebook_lab.query import count, recover_count, render_count
from tools.local_experiment.notebook.client import NotebookError
```

```python
table = "smoke.fixture_v1"  # catalog は iceberg 固定
request = "全行の件数を数えてください"  # 条件付きの例: idが1の行の件数を数えてください
```

```python
result = None
try:
    result = count(table, request)
except NotebookError as error:
    print("集計を完了できませんでした。固定エラーコード:", str(error))
    print("未確認の実行は自動再試行されません。専用 queries ディレクトリを確認してください。")
else:
    display(HTML(render_count(result)))
```

## 保存済み結果の再表示
下のセルに完了済み実行の token を指定します。保存した証跡を確認して再表示します。Qwen やテーブルへの追加読取は実行しません。表示する snapshot は集計時点のもので、最新件数の再取得ではありません。

```python
recovery_attempt = ""
saved_result = None
if recovery_attempt:
    try:
        saved_result = recover_count(recovery_attempt)
    except NotebookError as error:
        print("保存結果を確認できませんでした。固定エラーコード:", str(error))
    else:
        display(HTML(render_count(saved_result)))
```
